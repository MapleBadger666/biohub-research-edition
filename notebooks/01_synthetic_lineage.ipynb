# Synthetic lineage reconstruction

Install the edition package before opening this notebook. It generates its own example inputs. Set `BIOHUB_DATA` and `BIOHUB_WEIGHTS` for local NumPy inputs, and `BIOHUB_PREDICTIONS_DIR` and `BIOHUB_OUTPUT_DIR` to keep outputs in chosen directories. Otherwise outputs stay in a kernel-local temporary directory.


```python
import os
from pathlib import Path
from tempfile import TemporaryDirectory

from research_edition.demo import run

workspace = TemporaryDirectory(prefix="biohub-research-")
root = Path(workspace.name)
data = os.environ.get("BIOHUB_DATA")
weights = os.environ.get("BIOHUB_WEIGHTS")
result = run(
    Path(os.environ.get("BIOHUB_PREDICTIONS_DIR", root / "predictions")),
    Path(os.environ.get("BIOHUB_OUTPUT_DIR", root / "reports")),
    data=Path(data) if data else None,
    weights=Path(weights) if weights else None,
)
result
```
